## Goal · short exercise

Estimate how precision and context length affect memory. These are **weights-only** and **KV-cache** approximations, not a promise that a model will fit a particular GPU.


```python
parameters=4_000_000_000  # round illustrative count
for name,bytes_per_value in [('FP32',4),('FP16/BF16',2),('8-bit',1),('4-bit',.5)]:
    print(name,round(parameters*bytes_per_value/2**30,2),'GiB idealized weights')
layers,kv_heads,head_dim,batch=36,8,128,1
for length in [512,2048,8192]:
    gib=2*layers*kv_heads*head_dim*length*batch*2/2**30  # K and V, 2 bytes each
    print(length,'tokens →',round(gib,3),'GiB approximate KV cache')
```
