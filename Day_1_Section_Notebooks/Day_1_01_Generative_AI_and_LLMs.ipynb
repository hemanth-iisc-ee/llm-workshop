## Goal

Distinguish a classifier (predict a label) from a simple generative model (sample a new observation); then see next-token prediction as repeated conditional choices. Transformer configurations are a conceptual overview: BERT/RoBERTa are encoder-only; T5/BART are encoder–decoder; GPT-2/Llama/Qwen3 are decoder-only.


```python
import numpy as np
rng=np.random.default_rng(7)
observations=np.array([-2.2,-1.8,-1.3,1.2,1.7,2.4])
labels=(observations>0).astype(int)
def classify(x):return 'class B' if x>0 else 'class A'
mean,std=float(observations.mean()),float(observations.std(ddof=1))
samples=rng.normal(mean,std,size=5)
print('Classifier on x=1.5:',classify(1.5))
print('Illustrative generative samples:',np.round(samples,2))
print('A single normal distribution is deliberately a poor fit to two clusters.')
```

## Next-token loop

This tiny hand-written transition table is an analogy, not an LLM. A real language model learns conditional distributions from data.


```python
transitions={'the':{'battery':.7,'device':.3},'battery':{'lasts':.8,'charges':.2},'lasts':{'long':1.0}}
current='the';sentence=[current]
for _ in range(3):
    choices=transitions.get(current)
    if not choices:break
    current=rng.choice(list(choices),p=list(choices.values()))
    sentence.append(current)
print(' '.join(sentence))
# Exercise: add a transition for 'device' and sample ten continuations.
```
